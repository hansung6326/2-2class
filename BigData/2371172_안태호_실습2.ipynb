## 1번

### 1-1

```python
september_expenses = [300000, 50000, 70000, 30000]
print(f"리스트 전체 출력: {september_expenses}")
```

### 1-2

```python
september_expenses.append(400000)
september_expenses[1] = 60000
print(f"리스트 전체 출력: {september_expenses}")
```

### 1-3

```python
september_expenses.remove(30000)
print(f"리스트 전체 출력: {september_expenses}")
```

### 1-4

```python
print(f"도서비 삭제 여부: {30000 in september_expenses}, (True=존재함, False=존재하지 않음)")
print(f"전체 지출 항목의 개수: {len(september_expenses)}")
```

### 1-5

```python
september_expenses.sort()
print(f"오름차순 정렬 {september_expenses}")
september_expenses.sort(reverse=True)
print(f"내림차순 정렬 {september_expenses}")
```

### 1-6

```python
september_expenses += [150000, 80000]
print(f"리스트 전체 출력: {september_expenses}")
```

## 2번

### 2-1

```python
items = [("음료", 5000), ("간식", 3000), ("장식품", 30000)]
print(f"items 전체 출력: {items}")
```

### 2-2

```python
name, price = items[0]
print(f"{name} : {price}")
```

### 2-3

```python
try:
    items[0][1] = 6000
except TypeError as e:
    print(f"튜플 값 변경 불가: {e}")
```

### 2-4

```python
items[0] = ("음료", 6000)
print(f"items 전체 출력: {items}")
```

### 2-5

```python
items.append(("현수막", 20000))
items.remove(("간식", 3000))
print(f"items 전체 출력: {items}")
```

### 2-6

```python
sorted_items = sorted(items, key=lambda x: x[1])
print(f"items 전체 출력: {sorted_items}")
```
