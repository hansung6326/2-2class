## 1번

```python
def fahr2celsius(temp):
    celsius = (temp - 32) * (5 / 9)
    return celsius
```

### 1-1

```python
def fahr2kelvin(temp):
    kelvin = fahr2celsius(temp) + 273.15
    return kelvin
```

### 1-2

```python
def celsius2fahr(temp):
    fahr = (temp * 1.8) + 32
    return fahr
```

### 1-3

```python
def cel2FK(temp):
    fahr = celsius2fahr(temp)
    kelvin = temp + 273.15
    FK = {"fahrenheit": fahr, "kelvin": kelvin}
    return FK
```

### 실행

```python
print(cel2FK(0))
```

## 2번

```python
import math


def bday(n):
    nums = []
    for i in range(n):
        nums.append((365 - i) / 365)
    return 1 - math.prod(nums)
```

### 실행

```python
print(bday(2))
```
